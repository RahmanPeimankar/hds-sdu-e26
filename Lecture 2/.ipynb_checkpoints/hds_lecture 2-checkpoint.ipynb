# What is PhysioNet?

PhysioNet is a research resource for physiological and clinical data, software, and challenges.

Examples include:

- ECG;
- EEG and sleep;
- PPG;
- respiration;
- blood pressure;
- accelerometry;
- ICU waveforms;
- clinical variables;
- challenge datasets.

### Course principle
Whenever possible, project groups should start with a well-documented **Open Access** dataset.

# Why emphasise Open Access datasets?

For the semester project, Open Access is attractive because:

- every group member can access the data;
- another group can reproduce the work;
- your examiner can inspect the source;
- no credentialing delay is needed;
- access instructions are easier to document.

### Important
Open Access does **not** mean “upload the full raw dataset to GitHub”.

Always read the dataset page, licence, citation instructions, and version information.

# How to read a PhysioNet dataset page

Before writing code, identify:

1. What was measured?
2. Who or what was recorded?
3. How many records/participants are included?
4. Which channels/modalities are available?
5. What is the sampling frequency?
6. Which labels or annotations exist?
7. What is the access type?
8. Which publication should be cited?
9. Which version are you using?

### Rule
Read the documentation before start working with the data.

# The header is essential metadata

A WFDB header can tell us:

- record name;
- number of signals;
- sampling frequency;
- number of samples;
- signal names;
- physical units;
- gain/conversion information;
- comments.

### Before processing a signal, ask:
Do I know its **sampling frequency, channel name, and unit**?

# Discover records first

WFDB can read the record list from PhysioNet:

```python
records = wfdb.get_record_list("mitdb")
```

Why?

- record names may not be sequential;
- different databases use different naming conventions;
- the official record list is part of the dataset structure.

````python
try:
    records = wfdb.get_record_list("mitdb")
    print("Number of records:", len(records))
    print("First 10 record names:")
    print(records[:10])
except Exception as e:
    print("Could not retrieve the record list.")
    print("Error:", e)
````

# `rdheader()` — inspect before loading

```python
header = wfdb.rdheader(
    "100",
    pn_dir="mitdb"
)
```

Useful attributes:

```python
header.fs
header.n_sig
header.sig_len
header.sig_name
header.units
```

Metadata first, waveform second.

````python
try:
    header = wfdb.rdheader("100", pn_dir="mitdb")

    header_info = pd.Series({
        "record_name": header.record_name,
        "sampling_frequency_hz": header.fs,
        "number_of_channels": header.n_sig,
        "number_of_samples": header.sig_len,
        "signal_names": header.sig_name,
        "units": header.units
    })

    display(header_info)
except Exception as e:
    print("Could not read the header.")
    print("Error:", e)
````

````python
try:
    duration_seconds = header.sig_len / header.fs
    print(f"Duration: {duration_seconds:.1f} seconds")
    print(f"Duration: {duration_seconds/60:.1f} minutes")
except Exception:
    print("Run the header cell first.")
````

# `rdrecord()` — load only what you need

Instead of loading the full recording:

```python
record = wfdb.rdrecord(
    "100",
    pn_dir="mitdb",
    sampfrom=0,
    sampto=10 * int(header.fs)
)
```

`record.p_signal` contains physical signal values as a NumPy array.

### Data-science habit
Start with a small segment and verify everything before scaling up.

````python
try:
    fs = int(header.fs)

    record = wfdb.rdrecord(
        "100",
        pn_dir="mitdb",
        sampfrom=0,
        sampto=10*fs
    )

    print("Signal array shape:", record.p_signal.shape)
    print("Channels:", record.sig_name)
    print("Units:", record.units)
    print("Loaded duration:", record.p_signal.shape[0] / record.fs, "s")

except Exception as e:
    print("Could not load the signal.")
    print("Error:", e)
````

# Interpret the array shape

If:

```text
record.p_signal.shape = (3600, 2)
```

then:

- 3600 rows = time samples;
- 2 columns = channels.

At 360 Hz:

3600 / 360 = 10 seconds

### Connection to Lecture 1
A real WFDB waveform becomes a familiar NumPy array.

````python
try:
    t = np.arange(record.p_signal.shape[0]) / record.fs

    fig, axes = plt.subplots(record.n_sig, 1, figsize=(13, 6), sharex=True)
    if record.n_sig == 1:
        axes = [axes]

    for i, ax in enumerate(axes):
        ax.plot(t, record.p_signal[:, i], linewidth=0.9)
        ax.set_ylabel(f"{record.sig_name[i]}\n[{record.units[i]}]")
        ax.grid(alpha=0.25)

    axes[-1].set_xlabel("Time (s)")
    fig.suptitle("MIT-BIH record 100 — first 10 seconds")
    fig.tight_layout()
    plt.show()

except Exception as e:
    print("Run the record-loading cells first.")
    print("Error:", e)
````

# Select channels explicitly

For a multichannel record:

```python
record = wfdb.rdrecord(
    "100",
    pn_dir="mitdb",
    channels=[0]
)
```

Why select channels?

- reduce memory;
- simplify plots;
- focus on the signal needed for the question;
- avoid silently using the wrong column.

Always check `sig_name` and `units`.

# Annotations

A waveform tells us what was measured.

Annotations add information about **events or labels**.

```python
ann = wfdb.rdann(
    "100",
    "atr",
    pn_dir="mitdb"
)
```

Important fields:

```python
ann.sample
ann.symbol
```

Each annotation is linked to a sample position.

````python
try:
    ann = wfdb.rdann(
        "100",
        "atr",
        pn_dir="mitdb",
        sampfrom=0,
        sampto=10*fs
    )

    print("Number of annotations:", len(ann.sample))
    print("First sample positions:", ann.sample[:10])
    print("First annotation symbols:", ann.symbol[:10])

except Exception as e:
    print("Could not read annotations.")
    print("Error:", e)
````

# Sample number is not time

If an annotation is at sample 720 and:

```text
fs = 360 Hz
```

then:

720 / 360 = 2 seconds


In Python:

```python
annotation_time = ann.sample / fs
```

This conversion is essential when aligning events with a plot.

````python
try:
    ann_df = pd.DataFrame({
        "sample": ann.sample,
        "time_s": ann.sample / fs,
        "symbol": ann.symbol
    })

    display(ann_df.head(12))
except Exception:
    print("Run the annotation cell first.")
````

````python
try:
    signal = record.p_signal[:, 0]
    t = np.arange(len(signal)) / record.fs

    valid = ann.sample < len(signal)
    ann_samples = ann.sample[valid]

    plt.figure(figsize=(13, 4.5))
    plt.plot(t, signal, linewidth=1, label=record.sig_name[0])
    plt.scatter(
        ann_samples / fs,
        signal[ann_samples],
        marker="v",
        s=45,
        label="annotation"
    )

    plt.xlabel("Time (s)")
    plt.ylabel(f"{record.sig_name[0]} [{record.units[0]}]")
    plt.title("MIT-BIH record 100 — waveform with annotations")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.show()
except Exception as e:
    print("Run the signal and annotation cells first.")
    print("Error:", e)
````

````python
try:
    ann_df["interval_s"] = ann_df["time_s"].diff()
    display(ann_df.head(12))

    plt.figure(figsize=(10, 4))
    plt.plot(ann_df["time_s"], ann_df["interval_s"], marker="o")
    plt.xlabel("Annotation time (s)")
    plt.ylabel("Interval from previous annotation (s)")
    plt.title("Intervals between annotated events")
    plt.grid(alpha=0.25)
    plt.show()
except Exception:
    print("Run the annotation DataFrame cell first.")
````

# Common mistakes with real waveforms

1. Treating sample numbers as seconds  
2. Ignoring units  
3. Assuming column 0 is always the desired signal  
4. Loading the complete database immediately  
5. Losing record/participant identifiers  
6. Guessing annotation meanings  

### General rule
Metadata and provenance are part of the data.

# Streaming versus local download

### Streaming

```python
wfdb.rdrecord("100", pn_dir="mitdb")
```

Useful for exploration and small subsets.

### Local copy
Useful when repeatedly processing many records or when internet access is unreliable.

Whichever approach you choose, document it in:

```text
data/README.md
```

````python
def load_physionet_segment(
    database,
    record_name,
    start_s=0,
    duration_s=10,
    channels=None
):
    header = wfdb.rdheader(record_name, pn_dir=database)
    fs = header.fs

    sampfrom = int(start_s * fs)
    sampto = int((start_s + duration_s) * fs)

    record = wfdb.rdrecord(
        record_name,
        pn_dir=database,
        sampfrom=sampfrom,
        sampto=sampto,
        channels=channels
    )

    return record

# Example:
# segment = load_physionet_segment("mitdb", "100", start_s=5, duration_s=10)
````
