# `rdrecord()` returns a WFDB Record object

A `Record` object is more than an array.

It contains:

- `p_signal`: physical signal values;
- `fs`: sampling frequency;
- `sig_name`: channel names;
- `units`: physical units;
- `sig_len`: number of samples;
- `comments`: record comments;
- technical fields such as gain, baseline and format.

This is why we should not immediately extract only `p_signal` and discard the rest.

````python
# Read a short segment from MIT-BIH record 100 as a Record object

record = wfdb.rdrecord(
    "100",
    pn_dir="mitdb",
    sampfrom=0,
    sampto=10 * 360
)

print("Record name:", record.record_name)
print("Sampling frequency:", record.fs)
print("Signal shape:", record.p_signal.shape)
print("Signal names:", record.sig_name)
print("Units:", record.units)
print("Comments:", record.comments)
````

````python
# Inspect selected Record attributes as a table

record_info = pd.Series({
    "record_name": record.record_name,
    "n_sig": record.n_sig,
    "fs": record.fs,
    "sig_len_loaded": record.sig_len,
    "sig_name": record.sig_name,
    "units": record.units,
    "file_name": record.file_name,
    "fmt": record.fmt,
    "adc_gain": record.adc_gain,
    "baseline": record.baseline
})

display(record_info)
````

# Why inspect gain and baseline?

WFDB records often store digital samples and convert them to physical values.

Conceptually:

```text
digital value → gain/baseline correction → physical value
```

For most course exercises we use:

```python
record.p_signal
```

because it is already in physical units.

But knowing that conversion exists helps you understand why headers matter.

# `plot_wfdb()` for fast biosignal inspection

WFDB provides plotting helpers.

```python
wfdb.plot_wfdb(record=record, time_units="seconds")
```

This is useful for quickly checking:

- channel names;
- units;
- time scale;
- whether the loaded segment looks plausible;
- whether a channel appears flat, noisy or saturated.

Later, Matplotlib gives more control for publication-ready plots.

````python
wfdb.plot_wfdb(
    record=record,
    title="MIT-BIH record 100 — first 10 seconds",
    time_units="seconds"
)
````

# `rdsamp()` returns a NumPy array and a metadata dictionary

Sometimes you do not need the full Record object.

```python
signals, fields = wfdb.rdsamp(...)
```

Returns:

- `signals`: NumPy array;
- `fields`: dictionary with metadata.

This is convenient when building machine-learning-style pipelines.

````python
signals, fields = wfdb.rdsamp(
    "100",
    pn_dir="mitdb",
    sampfrom=0,
    sampto=10 * 360,
    channels=[0]
)

print("signals shape:", signals.shape)
display(pd.Series(fields))
````

# Intentional channel selection

A record may contain multiple channels or modalities.

Examples:

- ECG lead II;
- ECG V lead;
- PPG / PLETH;
- blood pressure;
- respiration;
- accelerometry.

You should choose channels based on the research question.

Never assume column 0 is automatically the correct signal.

````python
# Example inspired by the WFDB demo: Challenge 2015 record with ECG and PLETH

challenge_record = wfdb.rdrecord(
    "a103l",
    pn_dir="challenge-2015/training/",
    sampfrom=0,
    sampto=2500
)

print("Sampling frequency:", challenge_record.fs)
print("Signal names:", challenge_record.sig_name)
print("Units:", challenge_record.units)
print("Comments:", challenge_record.comments)

wfdb.plot_wfdb(
    record=challenge_record,
    title="Challenge 2015 a103l — ECG and PLETH example",
    time_units="seconds"
)
````

# What to observe in a multimodal record

In the previous example, the channels are not simply duplicates.

Possible questions:

- Which channels are ECG?
- Which channel is PLETH/PPG-like?
- Do the signals have comparable units?
- Do peaks occur with similar timing?
- Is one signal noisier than another?
- Could one modality help validate the other?

This is the beginning of multimodal biosignal thinking.

````python
# Example inspired by WFDB demo: selected channels from a multi-lead ECG record

ptb_signals, ptb_fields = wfdb.rdsamp(
    "s0010_re",
    channels=[14, 0, 5, 10],
    sampfrom=100,
    sampto=5000,
    pn_dir="ptbdb/patient001/"
)

print("PTB signal shape:", ptb_signals.shape)
display(pd.Series(ptb_fields))

t_ptb = np.arange(ptb_signals.shape[0]) / ptb_fields["fs"]

fig, axes = plt.subplots(ptb_signals.shape[1], 1, figsize=(13, 8), sharex=True)

for i, ax in enumerate(axes):
    ax.plot(t_ptb, ptb_signals[:, i], linewidth=0.8)
    ax.set_ylabel(f"{ptb_fields['sig_name'][i]}\n[{ptb_fields['units'][i]}]")
    ax.grid(alpha=0.25)

axes[-1].set_xlabel("Time (s)")
fig.suptitle("PTB multi-lead ECG — selected channels")
fig.tight_layout()
plt.show()
````

# Convert signals into pandas for exploration

A signal array is useful for numerical processing.

A DataFrame is useful for inspection and joining.

For a loaded record:

```python
record.to_dataframe()
```

can create a time-indexed DataFrame.

For simple teaching examples, we can also build it manually.

````python
# Manual conversion to a DataFrame for custom analysis

time = np.arange(record.p_signal.shape[0]) / record.fs

signal_df = pd.DataFrame(
    record.p_signal,
    columns=record.sig_name
)

signal_df.insert(0, "time_s", time)
signal_df.head()
````

````python
# Per-channel descriptive summaries

channel_summary = signal_df[record.sig_name].agg(
    ["mean", "std", "min", "max"]
).T

channel_summary["amplitude_range"] = channel_summary["max"] - channel_summary["min"]

display(channel_summary)
````

# Biosignal EDA questions for raw waveforms

For each channel, ask:

- Is the signal flat?
- Are there missing values?
- Is the amplitude plausible?
- Is there baseline drift?
- Are peaks visible?
- Are channels similar or very different?
- Does a short segment represent the whole record?

EDA does not solve these problems, but it reveals where to look.

````python
# Simple custom visualisation: same ECG channels overlaid after centering

centered = signal_df[record.sig_name] - signal_df[record.sig_name].mean()

plt.figure(figsize=(13, 4))
for ch in record.sig_name:
    plt.plot(signal_df["time_s"], centered[ch], label=f"{ch} centered", linewidth=0.9)

plt.xlabel("Time (s)")
plt.ylabel("Centered amplitude")
plt.title("Comparing ECG channels after centering")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
````

# Annotations make the waveform interpretable

Annotations link events to sample positions.

For MIT-BIH:

```python
ann = wfdb.rdann("100", "atr", pn_dir="mitdb")
```

Important fields:

- `ann.sample`: sample indices;
- `ann.symbol`: annotation symbols;
- `ann.aux_note`: optional notes;
- `ann.fs`: sampling frequency, if available or assigned.

````python
ann = wfdb.rdann(
    "100",
    "atr",
    pn_dir="mitdb",
    sampfrom=0,
    sampto=10 * 360
)

print("Number of annotations:", len(ann.sample))
print("First samples:", ann.sample[:10])
print("First symbols:", ann.symbol[:10])

ann_df = pd.DataFrame({
    "sample": ann.sample,
    "time_s": ann.sample / record.fs,
    "symbol": ann.symbol
})

display(ann_df.head())
````

# Plot signal and annotation together

This is one of the most important ECG exploration steps.

It helps answer:

- Do annotations align with visible peaks?
- Are there unexpected symbols?
- Is the selected segment suitable for analysis?
- Are there errors in sample-to-time conversion?

````python
wfdb.plot_wfdb(
    record=record,
    annotation=ann,
    title="MIT-BIH record 100 — signal with beat annotations",
    time_units="seconds"
)
````

````python
# Custom Matplotlib overlay for full control

sig0 = record.p_signal[:, 0]
t = np.arange(len(sig0)) / record.fs

valid = ann.sample < len(sig0)
ann_samples = ann.sample[valid]

plt.figure(figsize=(13, 4.5))
plt.plot(t, sig0, linewidth=1, label=record.sig_name[0])
plt.scatter(
    ann_samples / record.fs,
    sig0[ann_samples],
    marker="v",
    s=45,
    label="annotation"
)

plt.xlabel("Time (s)")
plt.ylabel(f"{record.sig_name[0]} [{record.units[0]}]")
plt.title("Custom annotation overlay")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
````

# First biosignal feature: RR intervals

For ECG beat annotations, consecutive beat times can produce RR intervals.

```python
RR_i = t_i - t_{i-1}
```

RR intervals are the basis for:

- heart-rate estimation;
- rhythm irregularity;
- HRV features;
- AF-related analysis.

Today we only compute and visualise them.

````python
ann_df["rr_interval_s"] = ann_df["time_s"].diff()
display(ann_df.head(10))

plt.figure(figsize=(10, 4))
plt.plot(ann_df["time_s"], ann_df["rr_interval_s"], marker="o")
plt.xlabel("Annotation time (s)")
plt.ylabel("RR interval (s)")
plt.title("RR intervals in the first 10 seconds")
plt.grid(alpha=0.25)
plt.show()
````

````python
# Convert RR intervals to approximate instantaneous heart rate

ann_df["instant_hr_bpm"] = 60 / ann_df["rr_interval_s"]

plt.figure(figsize=(10, 4))
plt.plot(ann_df["time_s"], ann_df["instant_hr_bpm"], marker="o")
plt.xlabel("Annotation time (s)")
plt.ylabel("Instantaneous HR (bpm)")
plt.title("Approximate heart rate from annotation intervals")
plt.grid(alpha=0.25)
plt.show()

display(ann_df[["time_s", "symbol", "rr_interval_s", "instant_hr_bpm"]])
````

# From event table to record-level summary

A single record may produce many tables:

### Signal table
one row = one sample

### Annotation table
one row = one annotated event

### Record summary table
one row = one record

Today we will create a record-level table while preserving the connection to signals and annotations.

````python
def biosignal_summary(database, record_name, duration_s=30, channel=0):
    header = wfdb.rdheader(record_name, pn_dir=database)
    fs = int(header.fs)

    record = wfdb.rdrecord(
        record_name,
        pn_dir=database,
        sampfrom=0,
        sampto=duration_s * fs,
        channels=[channel]
    )

    ann = wfdb.rdann(
        record_name,
        "atr",
        pn_dir=database,
        sampfrom=0,
        sampto=duration_s * fs
    )

    x = record.p_signal[:, 0]
    rr = np.diff(ann.sample / fs)

    return {
        "record_id": record_name,
        "fs": header.fs,
        "channel": record.sig_name[0],
        "unit": record.units[0],
        "duration_s_loaded": duration_s,
        "signal_mean": float(np.nanmean(x)),
        "signal_std": float(np.nanstd(x)),
        "signal_range": float(np.nanmax(x) - np.nanmin(x)),
        "n_annotations": len(ann.sample),
        "mean_rr_s": float(np.nanmean(rr)) if len(rr) else np.nan,
        "std_rr_s": float(np.nanstd(rr)) if len(rr) else np.nan,
        "mean_hr_bpm": float(60 / np.nanmean(rr)) if len(rr) else np.nan
    }
````

````python
records = ["100", "101", "102", "103", "104"]

summary_rows = []
for r in records:
    try:
        summary_rows.append(biosignal_summary("mitdb", r, duration_s=30, channel=0))
    except Exception as e:
        print(f"Could not process {r}: {e}")

summary = pd.DataFrame(summary_rows)
display(summary)
````

````python
# Visualise record-level biosignal summaries

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].bar(summary["record_id"], summary["n_annotations"])
axes[0].set_xlabel("Record")
axes[0].set_ylabel("Annotations in first 30 s")
axes[0].set_title("Annotation count")

axes[1].bar(summary["record_id"], summary["mean_hr_bpm"])
axes[1].set_xlabel("Record")
axes[1].set_ylabel("Approx. mean HR (bpm)")
axes[1].set_title("Approximate HR from annotations")

for ax in axes:
    ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
````

# What makes this useful for biosignal processing?

This workflow gives us evidence for later decisions:

- Which channel should we analyse?
- Do annotations align with waveform peaks?
- Are RR intervals plausible?
- Is there visible noise or drift?
- Are some records very different?
- Which variables should be preserved before preprocessing?

This is more than data wrangling: it is signal-aware exploration.
