# Why use GitHub in this course?

A semester project is not only a final report.

It is a **developing scientific workflow**.

GitHub helps you:

- work as a group;
- keep track of changes;
- document decisions;
- share code with your teacher and examiner;
- prepare for reproducible research;
- prepare for Docker-based deployment later in the course.

### Key message

GitHub is not only for submitting the final code.

It is a record of how your project develops.

# What should your final repository answer?

By the end of the semester, a new reader should understand:

- What problem did you investigate?
- Which paper did you reproduce or follow?
- Which dataset did you use?
- How can the data be accessed?
- How did you process the data?
- Which features did you extract?
- Which analysis/model did you use?
- How did you evaluate the result?
- How can the project be reproduced?

This is why repository organisation matters from the beginning.

# Step 1 — Create or use a GitHub account

Each student should have a GitHub account.

Use an account that you expect to keep after the course.

# Step 2 — One repository per group

Only **one repository per group**.

Do not create one repository per student.

### Recommended repository name

```text
HDS-F26-GroupXX-ProjectName
```

Examples:

```text
HDS-F26-Group03-AF-Detection
HDS-F26-Group05-Wearable-Stress
HDS-F26-Group07-ECG-Signal-Quality
```

Avoid:

```text

final
test123
new-project-final-final
```

# Step 3 — Create the repository on GitHub

On GitHub:

1. Click **+** in the upper-right corner.
2. Select **New repository**.
3. Add the repository name.
4. Add a short project description.
5. Choose repository visibility according to course instructions.
6. Select **Add a README file**.
7. Select Python `.gitignore`.
8. Click **Create repository**.

### Recommended visibility

Private during development is often safest.

Do not make restricted or confidential data public.

# Step 4 — Invite collaborators

The repository owner should:

1. Open the repository.
2. Go to **Settings**.
3. Open **Collaborators**.
4. Choose **Add people**.
5. Search for each group member’s GitHub username.
6. Send the invitation.

Every invited student must accept the invitation.

### Verify
Each group member should open the repository in their own browser and confirm that they have access.

# Step 5 — What should the repository contain first?

After the first week, your repository should contain at least:

```text

```

Later it will grow to include:

```text

Do not worry if your first version is simple.

A repository is supposed to evolve.

# Step 7 — Do not upload the full dataset

For PhysioNet projects, your repository should usually **not** contain the full dataset.

Instead, create:

```text
data/README.md
```

and explain:

- dataset title;
- source URL;
- PhysioNet identifier;
- access type;
- citation;
- how your code loads the data.

### Why?

- avoids large repositories;
- respects dataset rules;
- makes the analysis reproducible without copying data;
- prevents accidental sharing of restricted data.

# Example `data/README.md`

```markdown

# Step 8 — Clone the repository

The repository on GitHub is the online version.

To work on your own computer, clone it.

On GitHub:

1. Click **Code**.
2. Choose HTTPS.
3. Copy the repository URL.

Then in a terminal:

```bash

cd YOUR_REPOSITORY_NAME
```

Example:

```bash

cd HDS-F26-Group03-AF-Detection
```

# The basic Git workflow

You will use these commands repeatedly:

```bash
git pull
git status
git add
git commit
git push
```

A typical cycle:

```text
PULL → WORK → SAVE → STATUS → ADD → COMMIT → PUSH
```

### Rule
Always pull before starting work.

# Useful command examples

```bash
# Check what has changed
git status

# Add one file
git add notebooks/01_data_exploration.ipynb

# Add all relevant changes
git add .

# Commit the change
git commit -m "Add initial data exploration notebook"

# Push to GitHub
git push
```

### Commit message rule
A commit message should describe what changed.

# Good and bad commit messages

| Good | Bad |
|---|---|
| Add initial ECG data exploration | update |
| Implement WFDB record loader | stuff |
| Add signal-quality visualisations | final |
| Fix participant-level split | final2 |
| Update README with dataset details | it works |

### Ask yourself
Would this commit message make sense to another group member next week?

# Jupyter notebooks and GitHub

GitHub can display `.ipynb` notebooks.

Before committing a notebook:

1. Restart the kernel.
2. Run all cells from top to bottom.
3. Check for errors.
4. Remove unnecessary debugging output.
5. Add Markdown explanations.
6. Make figures readable.
7. Save the notebook.
8. Commit it.

### A notebook should tell a story:
question → data → method → result → interpretation.

# Use `.gitignore`

Some files should not be committed.

Examples:

```text
__pycache__/
*.pyc
.ipynb_checkpoints/
.env
.venv/
venv/
.DS_Store
data/raw/
```

### Why ignore `data/raw/`?

Because complete datasets can be large, restricted, or unnecessary to duplicate.

The code and instructions should explain how to obtain the data.

# Never upload secrets or confidential data

Never commit:

- passwords;
- API keys;
- access tokens;
- private SSH keys;
- GitHub personal access tokens;
- identifiable patient data;
- restricted health data.

### Dangerous example

```python
TOKEN = "my-secret-token"
```

If you accidentally commit a secret, deleting the line later may not be enough because Git keeps history.

# Working together

At the beginning, it is acceptable to work directly on `main`.

Later, for larger changes, you may use branches.

Example:

```bash
git switch -c feature-ecg-preprocessing
```

Then:

```bash
git add .
git commit -m "Implement ECG preprocessing"
git push -u origin feature-ecg-preprocessing
```

A branch lets you develop something without immediately changing the stable version.

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 5))
ax.axis("off")

# main line
ax.plot([0.5, 10.5], [2.5, 2.5], linewidth=3)
for x, label in [(1.0, "start"), (4.0, "main"), (7.0, "main"), (10.0, "main")]:
    ax.scatter(x, 2.5, s=120)
    ax.text(x, 2.75, label, ha="center", fontsize=10)

# feature branch
ax.plot([3.0, 5.0, 7.2], [2.5, 3.6, 2.5], linewidth=2)
ax.scatter([4.2, 5.6], [3.15, 3.25], s=100)
ax.text(5.0, 3.8, "feature branch", ha="center", fontsize=12, fontweight="bold")

ax.text(0.6, 1.6, "main = stable project version", fontsize=12)
ax.text(0.6, 1.25, "branch = temporary workspace for a larger change", fontsize=12)
ax.set_xlim(0, 11)
ax.set_ylim(0.8, 4.2)
ax.set_title("Branches keep experiments separate from the stable project",
             fontsize=16, fontweight="bold")
plt.show()
````

# Troubleshooting

### I cannot push
Check that you accepted the collaborator invitation.

### Git asks for login
Use GitHub’s recommended authentication method. Password login from Git may not work.

### I changed the same file as another group member
Run `git pull` and resolve the conflict carefully.

### I committed the wrong file
Ask for help early.

### My notebook is huge
Clear unnecessary outputs before committing.
