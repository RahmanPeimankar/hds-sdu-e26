# Part 1 — NumPy and physiological time-series data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
fs = 100
duration = 20
time = np.arange(0, duration, 1/fs)
signal = (
    0.8 * np.sin(2 * np.pi * 1.2 * time)
    + 0.25 * np.sin(2 * np.pi * 2.4 * time)
    + 0.15 * np.random.randn(len(time))
)
print('Signal created successfully.')
```

## Task 1.1 — Understand the array

```python
print('Number of samples:', len(signal))
print('Shape:', signal.shape)
print('Data type:', signal.dtype)
print('Minimum:', signal.min())
print('Maximum:', signal.max())
print('Mean:', signal.mean())
print('Standard deviation:', signal.std())
```

### Interpretation
The signal contains 2000 samples because it represents 20 seconds sampled at 100 Hz: 20 × 100 = 2000.

## Task 1.2 — Work with time and sampling frequency

```python
samples_per_second = fs
samples_five_seconds = 5 * fs
start_index = 5 * fs
stop_index = 10 * fs
signal_5_10 = signal[start_index:stop_index]
time_5_10 = time[start_index:stop_index]

print('Samples in 1 second:', samples_per_second)
print('Samples in 5 seconds:', samples_five_seconds)
print('Start index:', start_index)
print('Stop index:', stop_index)
print('Length of signal_5_10:', len(signal_5_10))
```

## Task 1.3 — Visualise the signal

```python
plt.figure(figsize=(12,4))
plt.plot(time, signal)
plt.xlabel('Time (s)')
plt.ylabel('Signal amplitude')
plt.title('Complete 20-second physiological signal')
plt.grid(alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(time_5_10, signal_5_10)
plt.xlabel('Time (s)')
plt.ylabel('Signal amplitude')
plt.title('Physiological signal from 5 to 10 seconds')
plt.grid(alpha=0.3)
plt.show()
```

### Example answer
The shorter plot makes local signal details easier to inspect because fewer samples are compressed into the same figure width. This can help us see waveform shape, short artefacts, or other local patterns that may be difficult to notice when the complete recording is displayed at once.

# Part 2 — Working with tabular health data using pandas

```python
data = {
    'participant_id': ['P01','P02','P03','P04','P05','P06','P07','P08','P09','P10'],
    'age': [24,31,28,45,52,37,41,29,34,48],
    'sex': ['F','M','F','M','F','M','F','M','F','M'],
    'activity': ['Walking','Rest','Running','Walking','Rest','Running','Walking','Rest','Running','Walking'],
    'heart_rate': [96,68,145,101,72,152,92,65,148,105],
    'steps': [1850,120,3240,2010,90,3560,1760,150,3380,2200],
    'recording_minutes': [28,30,25,30,29,24,30,27,26,30],
    'signal_quality': [0.95,0.98,0.82,0.91,0.97,0.76,0.94,0.96,0.79,0.89]
}
df = pd.DataFrame(data)
df
```

# Part 3 — Inspect the dataset

```python
print('First five rows:')
display(df.head())
print('\nShape:', df.shape)
print('\nColumn names:')
print(df.columns.tolist())
print('\nData types:')
print(df.dtypes)
print('\nDescriptive statistics:')
display(df.describe())
print('\nMissing values:')
print(df.isna().sum())
```

### Example answer
Each row represents one participant-level observation in this simplified dataset. The row contains demographic information, the participant's activity condition, and summary measurements such as heart rate, steps, recording duration, and signal quality.

# Part 4 — Selecting and filtering data

## Task 4.1 — Participants older than 35

```python
older_participants = df[df['age'] > 35]
display(older_participants)
print('Number of participants:', len(older_participants))
```

## Task 4.2 — Participants whose heart rate is greater than 100

```python
high_hr = df[df['heart_rate'] > 100]
display(high_hr)
print('Number of participants:', len(high_hr))
```

## Task 4.3 — Participants performing `Running`

```python
running = df[df['activity'] == 'Running']
display(running)
print('Number of participants:', len(running))
```

## Task 4.4 — Participants with signal quality below 0.85

```python
low_quality = df[df['signal_quality'] < 0.85]
display(low_quality)
print('Number of participants:', len(low_quality))
```

# Part 5 — Grouping and summarising

```python
mean_hr = df.groupby('activity')['heart_rate'].mean()
mean_steps = df.groupby('activity')['steps'].mean()
mean_quality = df.groupby('activity')['signal_quality'].mean()
summary_table = pd.DataFrame({
    'mean_heart_rate': mean_hr,
    'mean_steps': mean_steps,
    'mean_signal_quality': mean_quality
})
display(summary_table)
```

### Example answers
1. **Running** has the highest average heart rate.  
2. **Rest** has the lowest average heart rate.  
3. Yes. In general, heart rate is expected to increase when physical activity becomes more intense.  
4. Running appears to have the lowest average signal quality in this small dataset.  
5. No. This dataset is too small to conclude that running causes poorer signal quality. Other factors such as sensor placement, participant differences, movement, and measurement conditions could also influence signal quality.

# Part 6 — Data visualisation

## Figure 1 — Heart rate by participant

```python
plt.figure(figsize=(10,4))
plt.bar(df['participant_id'], df['heart_rate'])
plt.xlabel('Participant')
plt.ylabel('Heart rate (bpm)')
plt.title('Heart rate by participant')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

## Figure 2 — Steps versus heart rate

```python
plt.figure(figsize=(8,5))
plt.scatter(df['steps'], df['heart_rate'], s=70)
plt.xlabel('Steps')
plt.ylabel('Heart rate (bpm)')
plt.title('Steps versus heart rate')
plt.grid(alpha=0.3)
plt.show()
```

### Example answer
The scatter plot suggests that participants with more steps tend to have higher heart rates in this small dataset. The relationship is not perfectly linear, but the points form a clear overall upward pattern.

## Figure 3 — Mean heart rate by activity

```python
activity_hr = df.groupby('activity')['heart_rate'].mean()
plt.figure(figsize=(7,4))
plt.bar(activity_hr.index, activity_hr.values)
plt.xlabel('Activity')
plt.ylabel('Mean heart rate (bpm)')
plt.title('Mean heart rate by activity')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

### Example answer
Figure 3 communicates the relationship between activity and heart rate most clearly because it directly compares the average heart rate for each activity category. Figure 1 focuses on individuals, while Figure 2 shows a relationship between steps and heart rate rather than directly comparing the three activity groups.

# Part 7 — Introduce a data-quality problem

```python
df_missing = df.copy()
df_missing.loc[2, 'heart_rate'] = np.nan
df_missing.loc[7, 'signal_quality'] = np.nan
display(df_missing)
```

```python
print('Missing values per column:')
print(df_missing.isna().sum())
print('\nRows containing at least one missing value:')
display(df_missing[df_missing.isna().any(axis=1)])
mean_hr_missing = df_missing['heart_rate'].mean()
print('\nMean heart rate:', mean_hr_missing)
```

### What does pandas do?
By default, `pandas.Series.mean()` ignores `NaN` values. Therefore, the mean heart rate is calculated using only the non-missing observations.

### Example answer
Missing values should be inspected before analysis because they may change the effective sample size and can influence statistics, plots, and later models. We also need to understand *why* values are missing, because missingness may itself be related to participant behaviour, sensor failure, or measurement quality.

# Part 8 — Create new variables

```python
df['recording_seconds'] = df['recording_minutes'] * 60
df['steps_per_minute'] = df['steps'] / df['recording_minutes']
display(df[['participant_id','activity','steps','recording_minutes','steps_per_minute']])
```

### Example answer
`steps_per_minute` adjusts the number of steps for differences in recording duration. Two participants may have different total step counts simply because one was recorded for longer, so a rate can make the comparison fairer.

# Part 9 — A small Health Data Science question

> **Does physical activity appear to be associated with heart rate in this small dataset?**

```python
analysis_table = (
    df.groupby('activity')
      .agg(
          participants=('participant_id','count'),
          mean_heart_rate=('heart_rate','mean'),
          std_heart_rate=('heart_rate','std'),
          mean_steps_per_minute=('steps_per_minute','mean')
      )
)
display(analysis_table)
```

```python
plt.figure(figsize=(7,4))
plt.bar(analysis_table.index, analysis_table['mean_heart_rate'])
plt.xlabel('Activity')
plt.ylabel('Mean heart rate (bpm)')
plt.title('Mean heart rate across activity groups')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
print('Overall mean heart rate:', df['heart_rate'].mean())
print('\nMean heart rate by activity:')
print(df.groupby('activity')['heart_rate'].mean())
```

### Example conclusion
In this small dataset, physical activity appears to be associated with heart rate. Participants in the rest group have the lowest mean heart rate, walking participants have intermediate values, and running participants have the highest mean heart rate. This pattern is consistent with the expectation that greater physical activity increases cardiovascular demand. The data also show that participants with more steps generally have higher heart rates. However, this dataset contains only ten participants and was created for teaching purposes, so we cannot draw strong scientific conclusions or claim that activity alone caused the differences. Participant age, fitness, measurement conditions, and other factors may also contribute. A larger study with more participants and repeated measurements would be needed for a more reliable conclusion.

### Example answers

**1. What is the difference between a NumPy array and a pandas DataFrame?**  
A NumPy array is primarily designed for efficient numerical computation and usually contains values of a common data type. A pandas DataFrame is a labelled two-dimensional table that can contain columns with different data types.

**2. When would you prefer a NumPy array?**  
For numerical operations on signals, matrices, and other homogeneous numerical data.

**3. When would you prefer a pandas DataFrame?**  
For participant tables, metadata, mixed data types, labelled columns, filtering, grouping, and data wrangling.

**4. Why is visualising health data important before performing further analysis?**  
Plots can reveal patterns, outliers, artefacts, missing sections, and unexpected values that may not be obvious from numerical summaries alone.

**5. Why should column names, units, participant IDs and other metadata be preserved?**  
They provide the context needed to interpret values correctly and help prevent mistakes when data are processed, merged, segmented, or analysed.

# Optional challenge

```python
activity_mapping = {'Rest': 0, 'Walking': 1, 'Running': 2}
df['activity_code'] = df['activity'].map(activity_mapping)

plt.figure(figsize=(7,4))
plt.scatter(df['activity_code'], df['heart_rate'], s=70)
plt.xticks([0,1,2], ['Rest','Walking','Running'])
plt.xlabel('Activity')
plt.ylabel('Heart rate (bpm)')
plt.title('Activity code versus heart rate')
plt.grid(alpha=0.3)
plt.show()
```

### Example answer
Representing activities as 0, 1, and 2 can be convenient for plotting or some computational tasks, but the numbers may incorrectly imply a precise numerical distance between categories. For example, the difference between Rest and Walking is not necessarily equivalent to the difference between Walking and Running. Therefore, categorical variables should usually be treated as categories rather than ordinary continuous numbers unless the numerical ordering has a justified meaning.
